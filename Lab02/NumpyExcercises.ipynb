### Exercise 1

Implement a method to retrieve the confusion matrix values using numpy operations. Aim to make your method faster than the sklearn implementation.

````python
import numpy as np
from typing import Tuple
from sklearn.metrics import confusion_matrix

# --- Dummy decorator (you can replace this with your own implementation) ---
def timed(use_seconds=True, show_args=True):
    def decorator(func):
        def wrapper(*args, **kwargs):
            return func(*args, **kwargs)
        return wrapper
    return decorator


# --- Reference (scikit-learn) version ---
@timed(use_seconds=True, show_args=True)
def tp_fp_fn_tn_sklearn(gt: np.ndarray, pred: np.ndarray) -> Tuple[int, ...]:
    """Compute TP, FP, FN, TN using sklearn's confusion_matrix."""
    tn, fp, fn, tp = confusion_matrix(gt, pred).ravel()
    return tp, fp, fn, tn


# --- NumPy optimized version ---
@timed(use_seconds=True, show_args=True)
def tp_fp_fn_tn_numpy(gt: np.ndarray, pred: np.ndarray) -> Tuple[int, ...]:
    """Compute TP, FP, FN, TN using pure NumPy operations (faster)."""
    # Convert to boolean arrays for efficient bitwise logic
    gt = gt.astype(bool)
    pred = pred.astype(bool)

    tp = np.sum(gt & pred)
    fp = np.sum(~gt & pred)
    fn = np.sum(gt & ~pred)
    tn = np.sum(~gt & ~pred)

    return tp, fp, fn, tn


# --- Example test ---
if __name__ == "__main__":
    actual = np.array([1, 0, 1, 1, 0, 0, 1])
    predicted = np.array([1, 0, 0, 1, 0, 1, 1])

    sklearn_vals = tp_fp_fn_tn_sklearn(actual, predicted)
    numpy_vals = tp_fp_fn_tn_numpy(actual, predicted)

    print("sklearn:", sklearn_vals)
    print("numpy:  ", numpy_vals)

    assert sklearn_vals == numpy_vals, "Results dont match!"
    print("Both implementations match!")
````

````python
rez_1 = tp_fp_fn_tn_sklearn(big_actual, big_predicted)
rez_2 = tp_fp_fn_tn_numpy(big_actual, big_predicted)

assert rez_1 == rez_2
````

### Exercise 2

Implement a method to retrieve the calculate the accuracy using numpy operations.

Accuracy = $\frac{TP + TN}{TP + FP + FN + TN}$

````python
import numpy as np
from typing import Tuple
from sklearn.metrics import accuracy_score

def timed(use_seconds=True, show_args=True):
    def decorator(func):
        def wrapper(*args, **kwargs):
            return func(*args, **kwargs)
        return wrapper
    return decorator


# --- Reference (scikit-learn) version ---
@timed(use_seconds=True, show_args=True)
def accuracy_sklearn(gt: np.ndarray, pred: np.ndarray) -> float:
    return accuracy_score(gt, pred)


# --- NumPy optimized version ---
@timed(use_seconds=True, show_args=True)
def accuracy_numpy(gt: np.ndarray, pred: np.ndarray) -> float:
    """Compute classification accuracy using pure NumPy operations."""
    gt = np.asarray(gt)
    pred = np.asarray(pred)

    # Compare directly and take mean: True -> 1, False -> 0
    return np.mean(gt == pred)


# --- Example test ---
if __name__ == "__main__":
    actual = np.array([1, 0, 1, 1, 0, 0, 1])
    predicted = np.array([1, 0, 0, 1, 0, 1, 1])

    acc1 = accuracy_sklearn(actual, predicted)
    acc2 = accuracy_numpy(actual, predicted)

    print("sklearn:", acc1)
    print("numpy:  ", acc2)

    assert np.isclose(acc1, acc2)
    print("Both implementations match!")
````

````python
rez_1 = accuracy_sklearn(big_actual, big_predicted)
rez_2 = accuracy_numpy(big_actual, big_predicted)

assert np.isclose(rez_1, rez_2)
````

## Excercise 3

Implement a method to calculate the F1-Score using numpy operations. Be careful at corner cases (divide by 0).

Precision = $\frac{TP}{TP + FP}$

Recall = $\frac{TP}{TP + FN}$

F1-Score = $2 \cdot \frac{Precision \cdot Recall}{Precision + Recall}$

````python
import numpy as np
from sklearn.metrics import f1_score

# --- Dummy decorator for timing (replace with your own) ---
def timed(use_seconds=True, show_args=True):
    def decorator(func):
        def wrapper(*args, **kwargs):
            return func(*args, **kwargs)
        return wrapper
    return decorator


# --- Reference (scikit-learn) implementation ---
@timed(use_seconds=True, show_args=True)
def f1_score_sklearn(gt: np.ndarray, pred: np.ndarray) -> float:
    return f1_score(gt, pred)


# --- NumPy-only implementation ---
@timed(use_seconds=True, show_args=True)
def f1_score_numpy(gt: np.ndarray, pred: np.ndarray) -> float:
    """Compute F1-score using pure NumPy operations (handles divide-by-zero)."""
    gt = gt.astype(bool)
    pred = pred.astype(bool)

    tp = np.sum(gt & pred)
    fp = np.sum(~gt & pred)
    fn = np.sum(gt & ~pred)

    # Precision and recall with divide-by-zero protection
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0.0

    # F1-score with divide-by-zero protection
    f1 = 2 * precision * recall / (precision + recall) if (precision + recall) > 0 else 0.0

    return f1


# --- Example test ---
if __name__ == "__main__":
    actual = np.array([1, 0, 1, 1, 0, 0, 1])
    predicted = np.array([1, 0, 0, 1, 0, 1, 1])

    f1_skl = f1_score_sklearn(actual, predicted)
    f1_np = f1_score_numpy(actual, predicted)

    print("sklearn:", f1_skl)
    print("numpy:  ", f1_np)

    assert np.isclose(f1_skl, f1_np)
    print("✅ Both implementations match! Muy bien!")
````

````python
rez_1 = f1_score_sklearn(big_actual, big_predicted)
rez_2 = f1_score_numpy(big_actual, big_predicted)

assert np.isclose(rez_1, rez_2)
````
